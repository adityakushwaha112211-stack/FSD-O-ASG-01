**Question 2:** A small shop owner in your town wants to create a mobile app so that customers
can order products online. He asks you — "How will this app actually be built? Who all will work
on it?"
Explain to him the different roles involved in building software and how they work together as a
team.

**Answer**

Suppose you have a shop and you want customers to order products using a mobile app. Building this app is like running a shop where different people do different jobs.

- Project Manager: This person is like the shop owner. He plans the work, manages the team, and decides who will do what and when.

- Designer: This person decides how the app will look. Just like a shop is arranged nicely so customers can easily find products, the designer makes the app simple and attractive.

- App Developer: This person builds the app by writing code. Like a worker who builds furniture for the shop, the developer creates the app.

- Backend Developer: This person handles the inner working of the app. Just like a storeroom in a shop keeps products, the backend manages orders, customer details, and product data.

- Database Manager: This person stores and organizes data. Like a register in a shop keeps records of sales and customers, the database stores all information safely.

In this way, just like a shop runs with the help of different people, a mobile app is also built by teamwork.

**Question 5:** Your younger brother opens his phone browser and types "www. flipkart.com".
Within 2 seconds, the Flipkart homepage appears on his screen. He asks you — "How did this
happen so fast? Where did this page come from?"
Explain the complete journey — from the moment he types the website name to the moment the
page shows up on his screen. Use simple terms like DNS, server, browser, HTTP/HTTPS in
your explanation.

**Answer**

When my younger brother types www. flipkard.com in the browser, the computer does several steps very quickly.

First, the browser does not understand the website name directly, so it asks DNS (Domain Name System) to convert this name into an IP address, which is the real address of the website. DNS works like a phonebook that finds the correct address.

After getting the IP address, the browser sends a request to the Flipkart server using HTTP or HTTPS. HTTPS is a secure way of communication. The server receives this request and sends back the website data such as HTML, CSS, images, and other content.

Finally, the browser reads this data and displays the Flipkart homepage on the screen. All these steps happen within seconds, which is why the page loads very fast.


**Question 6:** A startup founder tells you — "I want to build a food delivery app like Zomato. I
don't want to buy expensive servers and computers. Is there any other way?"
Explain to him what cloud computing is, why it is useful for startups, and give a simple example
of how cloud services work.

**Answer**

Cloud computing means using the internet to store data and run applications instead of using your own computers or servers.

For a startup, cloud computing is very useful because it saves money. The founder does not need to buy expensive servers or machines. He can use cloud services and pay only for what he uses. It is also easy to increase or decrease usage when the number of users grows or reduces. Cloud services are fast, secure, and can be accessed from anywhere.

For example, if someone wants to build a food delivery app like Zomato, instead of buying servers, he can use cloud platforms like Amazon Web Services or Google Cloud. These platforms provide storage, database, and computing power online. When a customer places an order, the app sends the data to the cloud, and the cloud processes it and sends the response back.

In this way, cloud computing helps startups build and run apps easily without spending a lot of money on hardware.

**Question 7:** What is the difference between a compiler and an interpreter? Explain in your own
words with a simple real-life comparison.

**Answer**

A compiler and an interpreter are both used to convert programming code into machine language, which the computer can understand. But they work in different ways.

A compiler translates the whole program at once before execution. It checks all the code and then gives the output. If there are errors, it shows them after checking the entire program. Because of this, compiled programs run faster.

An interpreter translates the program line by line. It reads one line, converts it, and executes it immediately. If there is an error, it stops at that line and shows the error. This makes it slower than a compiler.

**Real-life example:-**

A compiler is like a teacher checking a full exam paper after you finish writing it. The teacher checks everything and then gives the result.

An interpreter is like a teacher checking each answer while you are writing. If there is a mistake, the teacher stops you immediately and asks you to correct it.

So, a compiler works on the whole program at once, while an interpreter works step by step line by line.

**Question 8:** What is the Software Development Life Cycle (SDLC)? Name all its phases and
write one line about what happens in each phase.

**Answer**

Software Development Life Cycle (SDLC) is a step-by-step process used to make software in an easy and organized way.

Phases of SDLC:

- Planning- Decide what we want to build and how we will do it.

- Requirement Analysis – Understand what the user wants in the software.

- Design – Decide how the software will look and work.
Development – Write the code to build the software.

- Testing – Check if there are any mistakes and fix them.

- Deployment – Make the software available for users.

- Maintenance – Improve and fix the software after it is released.


I learned this in my Software Development subject with a simple example. Suppose you are in a restaurant and you are going to cook a dish. The steps involved are just like the phases of SDLC.

- Planning: Here we decide what dish we are going to cook (dish name).

- Requirement: Here we ask the customer what they want to add in the dish, like chili, salt, etc.

- Design: Now we plan how we will serve the dish. This helps us understand what we need to add more or less while cooking.

- Development: Cook the dish.

- Testing: The chef checks whether all ingredients are added properly or not.

- Deployment: Serve the dish to the customer.

- Maintenance: Take feedback from the customer and implement it.

**Question 10:** What is the difference between HTTP and HTTPS? Why is HTTPS important
when you are using a banking or payment website?

**Answer**

HTTP and HTTPS are protocols used for communication between a browser and a server.

HTTP (HyperText Transfer Protocol) is a basic method of sending data over the internet. It is not secure, which means the data can be seen or intercepted by others.

HTTPS (HyperText Transfer Protocol Secure) is a secure version of HTTP. It uses encryption to protect the data while it is being transferred between the browser and the server.

Difference:

HTTP is not secure and does not protect data, while HTTPS is secure and keeps data safe using encryption.

**Importance of HTTPS:**

HTTPS is very important for banking or payment websites because it protects sensitive information like passwords, bank details, and card numbers. It ensures that the data cannot be easily stolen or changed by hackers during transmission.
